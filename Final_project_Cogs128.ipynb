```python
commute_time_indicator_df = df[df['Indicator'] == 'Commute Time and Method of Transportation'].copy()
commute_method_distribution_df = commute_time_indicator_df[commute_time_indicator_df['Category'] == 'Commute Method of Transportation'].copy()

commute_method_distribution_df['Percent'] = pd.to_numeric(
    commute_method_distribution_df['Percent'].str.replace('%', ''), errors='coerce'
)

commute_method_distribution_df = commute_method_distribution_df.dropna(subset=['Percent'])
```

```python
print(commute_time_indicator_df['Category'].unique())
```

```python
specific_commute_methods = [
    'Private motor vehicle', 'Public transport', 'Bicycle', 'Walked only', 'Other', 'Worked at home',
    'Total Private Motor Vehicle', 'Total Public Transport', 'Private Motor Vehicle', 'Public Transport',
    'Walked Only', 'Worked at Home'
]

racial_ethnic_subcategories = [
    'American Indian or Alaska Native', 'Asian', 'Black or African American',
    'Hispanic or Latino, of Any Race', 'Native Hawaiian or Pacific Islander',
    'White', 'Multiracial', 'Some Other Race', 'Non-Hispanic Other Races',
    'Non-Hispanic White', 'Hispanic', 'Non-Hispanic Black'
]

# Filter for rows where Category is one of the specific commute methods
# AND Subcategory is one of the racial/ethnic groups
commute_method_by_race_df = commute_time_indicator_df[
    (commute_time_indicator_df['Category'].isin(specific_commute_methods)) &
    (commute_time_indicator_df['Subcategory'].isin(racial_ethnic_subcategories))
].copy()

# Convert 'Percent' to numeric, handling missing values
commute_method_by_race_df['Percent'] = pd.to_numeric(
    commute_method_by_race_df['Percent'].str.replace('%', ''), errors='coerce'
)

# Drop rows where 'Percent' is NaN after conversion
commute_method_by_race_df = commute_method_by_race_df.dropna(subset=['Percent'])

# Normalize the 'Category' column to handle different spellings/casing of commute methods
def normalize_commute_category(category):
    category = str(category).lower()
    if 'private motor vehicle' in category:
        return 'Private Motor Vehicle'
    elif 'public transport' in category:
        return 'Public Transport'
    elif 'walked only' in category:
        return 'Walked Only'
    elif 'worked at home' in category:
        return 'Worked at Home'
    elif 'bicycle' in category:
        return 'Bicycle'
    elif 'other' in category:
        return 'Other'
    return category

commute_method_by_race_df['Commute Method'] = commute_method_by_race_df['Category'].apply(normalize_commute_category)

print("Shape of commute_method_by_race_df: ", commute_method_by_race_df.shape)
print("First 5 rows of commute_method_by_race_df:")
print(commute_method_by_race_df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by 'Subcategory' and 'Commute Method', then sum the 'Percent' to get the total distribution
# This assumes 'Percent' values for each commute method within a subcategory can be aggregated this way.
# If the 'Percent' columns are meant to be proportions that sum to 100% per subcategory, we should ensure that first.
# For now, let's assume summing is appropriate for overall distribution across methods for each race.

# Ensure the 'Subcategory' column has a consistent set of racial/ethnic group names.
# Some subcategories might have similar meanings, e.g., 'Hispanic or Latino, of Any Race' and 'Hispanic'
# For now, I'll use the existing 'Subcategory' names as they are.

# Pivot the table to get commute methods as columns and subcategories as index
pivoted_commute_df = commute_method_by_race_df.pivot_table(
    index='Subcategory',
    columns='Commute Method',
    values='Percent',
    aggfunc='mean' # Use mean if multiple entries per Subcategory/Commute Method combination
).fillna(0) # Fill NaN values with 0, meaning 0% for that method/race combination

# Sort the index for better visualization
pivoted_commute_df = pivoted_commute_df.sort_index()

# Create the stacked bar chart
plt.figure(figsize=(14, 8))
pivoted_commute_df.plot(kind='bar', stacked=True, colormap='viridis', figsize=(14, 8))
plt.title('Distribution of Commute Methods by Race/Ethnicity')
plt.xlabel('Race/Ethnicity')
plt.ylabel('Percentage of Commuters')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Commute Method', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

print("Stacked bar chart showing commute method distribution by race/ethnicity has been generated.")
```

```python
specific_commute_methods = [
    'Private motor vehicle', 'Public transport', 'Bicycle', 'Walked only', 'Other', 'Worked at home',
    'Total Private Motor Vehicle', 'Total Public Transport', 'Private Motor Vehicle', 'Public Transport',
    'Walked Only', 'Worked at Home'
]

racial_ethnic_subcategories = [
    'American Indian or Alaska Native', 'Asian', 'Black or African American',
    'Hispanic or Latino, of Any Race', 'Native Hawaiian or Pacific Islander',
    'White', 'Multiracial', 'Some Other Race', 'Non-Hispanic Other Races',
    'Non-Hispanic White', 'Hispanic', 'Non-Hispanic Black'
]

# Filter for rows where 'Indicator' is 'Commute Time and Method of Transportation'
commute_time_indicator_df = df[df['Indicator'] == 'Commute Time and Method of Transportation'].copy()

# Filter for rows where Category is one of the specific commute methods
# AND Subcategory is one of the racial/ethnic groups
commute_method_by_race_df = commute_time_indicator_df[
    (commute_time_indicator_df['Category'].isin(specific_commute_methods)) &
    (commute_time_indicator_df['Subcategory'].isin(racial_ethnic_subcategories))
].copy()

# Convert 'Percent' to numeric, handling missing values
commute_method_by_race_df['Percent'] = pd.to_numeric(
    commute_method_by_race_df['Percent'].str.replace('%', ''), errors='coerce'
)

# Drop rows where 'Percent' is NaN after conversion
commute_method_by_race_df = commute_method_by_race_df.dropna(subset=['Percent'])

# Normalize the 'Category' column to handle different spellings/casing of commute methods
def normalize_commute_category(category):
    category = str(category).lower()
    if 'private motor vehicle' in category:
        return 'Private Motor Vehicle'
    elif 'public transport' in category:
        return 'Public Transport'
    elif 'walked only' in category:
        return 'Walked Only'
    elif 'worked at home' in category:
        return 'Worked at Home'
    elif 'bicycle' in category:
        return 'Bicycle'
    elif 'other' in category:
        return 'Other'
    return category

commute_method_by_race_df['Commute Method'] = commute_method_by_race_df['Category'].apply(normalize_commute_category)

# Pivot the table to get commute methods as columns and subcategories as index
pivoted_commute_df = commute_method_by_race_df.pivot_table(
    index='Subcategory',
    columns='Commute Method',
    values='Percent',
    aggfunc='mean' # Use mean if multiple entries per Subcategory/Commute Method combination
).fillna(0) # Fill NaN values with 0, meaning 0% for that method/race combination

# Sort the index for better visualization
pivoted_commute_df = pivoted_commute_df.sort_index()

print("Shape of pivoted_commute_df:", pivoted_commute_df.shape)
print("\nFirst 5 rows of pivoted_commute_df:")
display(pivoted_commute_df.head())
```
